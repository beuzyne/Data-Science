# Employee Salary & Tax Analysis with Pandas

```python
#calling pandas as a library
import pandas as pd

#prepare the data in dictionary
data_karyawan = {
    "Nama" : ["Max", "Mia", "Jacob"],
    "Posisi" : ["Junior DS", "Mid DS", "Senior DS"],
    "Gaji_Kotor" : [4500, 5000, 6000]
}

#DataFrame
df = pd.DataFrame(data_karyawan)

#run
print(df)
```

```python
#calculating the tax deduction column from the gross salary column
df["Potongan_Pajak"] = df["Gaji_Kotor"] *0.40

#Menghitung kolom gaji bersih
df["Gaji_Bersih"] = df["Gaji_Kotor"] - df ["Potongan_Pajak"]
```

```python
print (df)
```

```python
#Get annual bonus
df["bonus"] = df["Gaji_Kotor"] * 0.10

print(df)
```

```python
#filtering salary up to 5000
karyawan_gaji_tinggi = df[df["Gaji_Kotor"] > 5000]
print(karyawan_gaji_tinggi)
```

```python
#filtering salary less than or equal to €5000
employee_certain = df[df["Gaji_Kotor"] <= 5000]
print(employee_certain)
```

```python
#calling library for graphics
import matplotlib.pyplot as plt

#making bar chart from name and sallary
df.plot(x="Nama", y="Gaji_Kotor", kind="bar", color="purple")

#making title in bar chart
plt.title("Perbandingan Gaji Kotor Karyawan")
plt.ylabel("Gaji dalam Euro(€)")

#show
plt.show ()
```
